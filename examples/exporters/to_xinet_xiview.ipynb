# **Exporting Crosslinks to xiNET and xiVIEW**

```python
from pyXLMS import __version__

print(f"Installed pyXLMS version: {__version__}")
```

```python
from pyXLMS import parser
from pyXLMS import exporter
```

All exporting functionality is available via the `exporter` submodule. We also import the `parser` submodule to read crosslink-spectrum-matches and crosslinks.

```python
parser_result = parser.read(
    "../../data/ms_annika/XLpeplib_Beveridge_QEx-HFX_DSS_R1_Crosslinks.xlsx",
    engine="MS Annika",
    crosslinker="DSS",
)
xls = parser_result["crosslinks"]
```

We read crosslinks using the [generic parser](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.parser.html#pyXLMS.parser.read) from a single MS Annika `.xlsx` file. For easier access we also assign our crosslinks to the variable `xls`.

```python
from pyXLMS.transform import validate, targets_only

xls = targets_only(validate(xls))
```

**For down-stream analysis with xiNET or xiVIEW you would usually want to only consider high-confidence target crosslinks, either by pre-filtering in the search engine of your choice or via the pyXLMS functions `validate()` \[[docs](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.transform.html#pyXLMS.transform.validate.validate)\] and `targets_only()` \[[docs](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.transform.html#pyXLMS.transform.targets_only.targets_only)\]!**

```python
_df = exporter.to_xinet(xls, filename=None)
```

The function `exporter.to_xinet()` exports a list of crosslinks to [xiNET](https://crosslinkviewer.org/) format. The tool xiNET is accessible via the link [crosslinkviewer.org](https://crosslinkviewer.org/). Requires that `alpha_proteins`, `beta_proteins`, `alpha_proteins_crosslink_positions` and `beta_proteins_crosslink_positions` fields are set for all crosslinks. You can read more about the `to_xinet()` function and all its parameters here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.exporter.html#pyXLMS.exporter.to_xinet.to_xinet).

**Specifying `filename=None` will only return the [pandas](https://pandas.pydata.org/) `DataFrame` and not write it to disk!**

```python
_df = exporter.to_xiview(xls, filename=None)
```

The function `exporter.to_xiview()` exports a list of crosslinks to [xiVIEW](https://xiview.org/) format. The tool xiVIEW is accessible via the link [xiview.org](https://xiview.org/). Requires that `alpha_proteins`, `beta_proteins`, `alpha_proteins_crosslink_positions` and `beta_proteins_crosslink_positions` fields are set for all crosslinks. You can read more about the `to_xiview()` function and all its parameters here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.exporter.html#pyXLMS.exporter.to_xiview.to_xiview).

**Specifying `filename=None` will only return the [pandas](https://pandas.pydata.org/) `DataFrame` and not write it to disk!**
