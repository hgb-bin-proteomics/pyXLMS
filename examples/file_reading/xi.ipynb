```python
from pyXLMS import __version__

print(f"Installed pyXLMS version: {__version__}")
```

All functionality to parse crosslink-spectrum-matches (CSMs) and crosslinks (XLs) from xiSearch result files is available via the `parser` submodule. 

We also import the `transform` submodule to show some summary statistics of the read files.

```python
from pyXLMS import parser
from pyXLMS import transform
```

## **Reading xiSearch/xiFDR Result Files via `parser.read()`**

```python
parser_result_CSMs_FDR = parser.read(
    "../../data/xi/1perc_xl_boost_CSM_xiFDR2.2.1.csv",
    engine="xiSearch/xiFDR",
    crosslinker="BS3",
)
```

The xiFDR result file can be read using the `parser.read()` method and setting `engine="xiSearch/xiFDR"`. 

The method also requires us to specify crosslinker used for the experiment, which in this case is BS3 (`crosslinker="BS3"`).

 You can read the documentation for the `parser.read()` method here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.parser.html#pyXLMS.parser.read).

```python
for k, v in parser_result_CSMs_FDR.items():
    print(f"{k}: {type(v) if isinstance(v, list) else v}")
```

The `parser.read()` method returns a dictionary with a set of specified keys and their values. We refer to this dictionary as a `parser_result` object. 

All `parser.read*` methods return such a `parser_result` object, you can read more about that here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.html#pyXLMS.data.create_parser_result), and here: [**data types specification**](https://github.com/hgb-bin-proteomics/pyXLMS/blob/master/docs/data_types.md).

As you can see from the `parser_result` the ixFDR result file contains CSMs. See :(`crosslink-spectrum-matches: <class 'list'>`) in the print out. 

We would be able to access those via `parser_result["crosslink-spectrum-matches"]`. We will do this a bit further down. 

```python
_ = transform.summary(parser_result_CSMs_FDR)
```

With the `transform.summary()` method we can also print out some summary statistics about the CSMs in the file. You can read more about the method here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.transform.html#pyXLMS.transform.summary.summary).

```python
sample_csm = parser_result_CSMs_FDR["crosslink-spectrum-matches"][0]
for k, v in sample_csm.items():
    print(f"{k}: {v}")
```

Using `parser_result_CSMs_FDR["crosslink-spectrum-matches"][0]` we can get the first CSM of the file and take a closer look at that.  

This is an example CSM, you can learn more about the specific attributes and their values here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.html#pyXLMS.data.create_csm), and here: [**data types specification**](https://github.com/hgb-bin-proteomics/pyXLMS/blob/master/docs/data_types.md).

## **Reading xiSearch/xiFDR Result Files via `parser.read_xi()`**

```python
parser_result_CSMs_FDR = parser.read_xi(
    "../../data/xi/1perc_xl_boost_CSM_xiFDR2.2.1.csv"
)
```

We can also read xiSearch and xiFDR result files using the `parser.read_xi()` method which allows a more nuanced control over reading the result files - although everything theoretically can be done with the `parser.read()` function as well. You can read the documentation for the `parser.read_xi()` method here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.parser.html#pyXLMS.parser.parser_xldbse_xi.read_xi).

```python
_ = transform.summary(parser_result_CSMs_FDR)
```

```python
from pyXLMS.constants import XI_MODIFICATION_MAPPING

XI_MODIFICATION_MAPPING
```

******
By default the xiSearch/xiFDR parser considers all modifications that are in `constants.XI_MODIFICATION_MAPPING` as shown below for **pyXLMS version 1.4.2** . A full list of the default xi modifications is given here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.html#pyXLMS.constants.XI_MODIFICATION_MAPPING).

```python
my_mods = dict(XI_MODIFICATION_MAPPING)
my_mods["me"] = ('Methylation',14.01565)
my_mods
```

If you have any additional modifications in your result file(s) the parser needs to know about them, which is done via the `modifications` parameter that allows for passing a custom dictionary of modifications. 

It is usually a good idea to base this custom dictionary on `constants.XI_MODIFICATION_MAPPING` and add your modifications after, as shown here for methylation.

```python
parser_result_CSMs_FDR = parser.read_xi(
    "../../data/xi/1perc_xl_boost_CSM_xiFDR2.2.1.csv",
    modifications=my_mods
)
```

You can then pass the full list of expected modifications `my_mods` via the `modifications` parameter.

```python
_ = transform.summary(parser_result_CSMs_FDR)
```

******

There are several other parameters that can be set, you can read more about them here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.parser.html#pyXLMS.parser.parser_xldbse_xi.read_xi).
