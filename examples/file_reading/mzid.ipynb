# **Reading mzIdentML Result Files**

```python
from pyXLMS import __version__

print(f"Installed pyXLMS version: {__version__}")
```

```python
from pyXLMS import parser
from pyXLMS import transform
```

All functionality to parse crosslink-spectrum-matches (CSMs) and crosslinks (XLs) from mzIdentML result files is available via the `parser` submodule. We also import the `transform` submodule to show some summary statistics of the read files.

## **Reading mzIdentML Result Files via `parser.read()`**

```python
parser_result = parser.read(
    "../../data/ms_annika/XLpeplib_Beveridge_QEx-HFX_DSS_R1.mzid",
    engine="mzIdentML",
    crosslinker="DSS",
)
```

We can read any mzIdentML result file using the `parser.read()` method and setting `engine="mzIdentML"`. The method also requires us to specify the used crosslinker, in this case DSS was used (`crosslinker="DSS"`). You can read the documentation for the `parser.read()` method here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.parser.html#pyXLMS.parser.read).

```python
for k, v in parser_result.items():
    print(f"{k}: {type(v) if isinstance(v, list) else v}")
```

The `parser.read()` method returns a dictionary with a set of specified keys and their values. We refer to this dictionary as a `parser_result` object. All `parser.read*` methods return such a `parser_result` object, you can read more about that here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.html#pyXLMS.data.create_parser_result), and here: [**data types specification**](https://github.com/hgb-bin-proteomics/pyXLMS/blob/master/docs/data_types.md).

As you can see from the `parser_result` the mzIdentML result file contained CSMs. We would be able to access those via `parser_result["crosslink-spectrum-matches"]`.

```python
_ = transform.summary(parser_result)
```

With the `transform.summary()` method we can also print out some summary statistics about our read CSMs. You can read more about the method here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.transform.html#pyXLMS.transform.summary.summary).

```python
sample_csm = parser_result["crosslink-spectrum-matches"][0]
for k, v in sample_csm.items():
    print(f"{k}: {v}")
```

Here is an example CSM, you can learn more about the specific attributes and their values here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.html#pyXLMS.data.create_csm), and here: [**data types specification**](https://github.com/hgb-bin-proteomics/pyXLMS/blob/master/docs/data_types.md).

Similarly, with the `transform.summary()` method we can also print out some summary statistics about our read crosslinks.

In this example `parser_result["crosslinks"]` is `None`  because no valid crosslink result file is provided. Therefore, no crosslinks can be displayed here.


Here is an example crosslink, you can learn more about the specific attributes and their values here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.html#pyXLMS.data.create_crosslink), and here: [**data types specification**](https://github.com/hgb-bin-proteomics/pyXLMS/blob/master/docs/data_types.md).

```python
parser_result = parser.read(
    "../../data/ms_annika/XLpeplib_Beveridge_QEx-HFX_DSS_R1.mzid",
    engine="mzIdentML",
    crosslinker="DSS",
    parse_modifications=False,
)
```

We can also tell the parser to not parse modifications via `parse_modifications=False`, this might be useful if you don't care about post-translational-modifications, or if you have unknown modifications in your results that you would have to manually specify - _and you don't want to do that_.

In case you want to parse modifications but have unknown modifications in your results, you have to set them via the `modifications` parameter that can be passed via `**kwargs` to `parser.read_mzid()`. More about that later...

```python
sample_csm = parser_result["crosslink-spectrum-matches"][0]
for k, v in sample_csm.items():
    print(f"{k}: {v}")
```

Notice that this time the fields `alpha_modifications` and `beta_modifications` are empty (`None`) for our sample CSM in contrast to when we looked at it further up.

## **Reading mzIdentML Result Files via `parser.read_mzid()`**

```python
parser_result = parser.read_mzid(
    "../../data/ms_annika/XLpeplib_Beveridge_QEx-HFX_DSS_R1.mzid"
)
```

We can also read any mzIdentML result file using the `parser.read_mzid()` method which allows a more nuanced control over reading mzIdentML result files - even though theoretically everything can be done with the `parser.read()` function as well. You can read the documentation for the `parser.read_mzid()` method here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.parser.html#pyXLMS.parser.parser_xldbse_mzid.read_mzid).

```python
_ = transform.summary(parser_result)
```

*****

```python
from pyXLMS.constants import CROSSLINKERS

CROSSLINKERS
```

By default the mzIdentML parser considers all crosslinkert that are in `constants.CROSSLINKERS` as shown above for **pyXLMS version 1.4.3** - a full list of default crosslinkers is given here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.html#pyXLMS.constants.CROSSLINKERS).

```python
my_cls = dict(CROSSLINKERS)
my_cls["EGS"] = 222.089
my_cls
```

If you have any additional crosslinkers in your result file(s) the parser needs to know about them, which is done via the `crosslinkers` parameter that allows for passing a custom dictionary of crosslinkers. It is usually a good idea to base this custom dictionary on `constants.CROSSLINKERS` and add your crosslinkers after, as shown above for EGS.

```python
parser_result = parser.read_mzid(
    "../../data/ms_annika/XLpeplib_Beveridge_QEx-HFX_DSS_R1.mzid",
    crosslinkers=my_cls,
)
```

You can then pass the full list of expected crosslinkers `my_cls` via the `crosslinkers` parameter.

```python
_ = transform.summary(parser_result)
```
