# **Reading Scout Result Files**

```python
from pyXLMS import __version__

print(f"Installed pyXLMS version: {__version__}")
```

```python
from pyXLMS import parser
from pyXLMS import transform
```

All functionality to parse crosslink-spectrum-matches (CSMs) and crosslinks (XLs) from Scout result files is available via the `parser` submodule. 

We also import the `transform` submodule to show some summary statistics of the read files.

## **Reading Scout Result Files via `parser.read()`**

```python
parser_result_Cas9_filtered_CSMs = parser.read(
    "../../data/scout/Cas9_Filtered_CSMs.csv",
    engine="Scout",
    crosslinker="DSSO",
)
```

The Scout result file can be read using the `parser.read()` method and setting `engine="Scout"`. 

The method also requires us to specify crosslinker used for the experiment, which in this case is DSSO (`crosslinker="DSSO"`).

 You can read the documentation for the `parser.read()` method here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.parser.html#pyXLMS.parser.read).

```python
for k, v in parser_result_Cas9_filtered_CSMs.items():
    print(f"{k}: {type(v) if isinstance(v, list) else v}")
```

The `parser.read()` method returns a dictionary with a set of specified keys and their values. We refer to this dictionary as a `parser_result` object. 

All `parser.read*` methods return such a `parser_result` object, you can read more about that here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.html#pyXLMS.data.create_parser_result), and here: [**data types specification**](https://github.com/hgb-bin-proteomics/pyXLMS/blob/master/docs/data_types.md).

As you can see from the `parser_result` the ixFDR result file contains CSMs. See :(`crosslink-spectrum-matches: <class 'list'>`) in the print out. 

We would be able to access those via `parser_result["crosslink-spectrum-matches"]`. We will do this a bit further down. 

```python
_ = transform.summary(parser_result_Cas9_filtered_CSMs)
```

With the `transform.summary()` method we can also print out some summary statistics about the CSMs in the file. You can read more about the method here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.transform.html#pyXLMS.transform.summary.summary).

```python
sample_csm = parser_result_Cas9_filtered_CSMs["crosslink-spectrum-matches"][0]
for k, v in sample_csm.items():
    print(f"{k}: {v}")
```

Using `parser_result_Cas9_filtered_CSMs["crosslink-spectrum-matches"][0]` we can get the first CSM of the file and take a closer look at that.  

This is an example CSM, you can learn more about the specific attributes and their values here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.html#pyXLMS.data.create_csm), and here: [**data types specification**](https://github.com/hgb-bin-proteomics/pyXLMS/blob/master/docs/data_types.md).

## **Reading Scout Result Files via `parser.read_scout()`**

```python
parser_result_Cas9_filtered_CSMs = parser.read_scout(
    "../../data/scout/Cas9_Filtered_CSMs.csv",
    crosslinker="DSSO"
)
```

We can also read Scout results files using the `parser.read_scout()` method which allows a more nuanced control over reading the result files - although everything theoretically can be done with the `parser.read()` function as well. You can read the documentation for the `parser.read_scout()` method here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.parser.html#pyXLMS.parser.parser_xldbse_scout.read_scout).

```python
_ = transform.summary(parser_result_Cas9_filtered_CSMs)
```

******

```python
from pyXLMS.constants import SCOUT_MODIFICATION_MAPPING

SCOUT_MODIFICATION_MAPPING
```

By default the Scout parser considers all modifications that are in `constants.SCOUT_MODIFICATION_MAPPING` as shown below for **pyXLMS version 1.4.2** . 

A full list of the default Scout modifications is given here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.html#pyXLMS.constants.SCOUT_MODIFICATION_MAPPING).

```python
my_mods = dict(SCOUT_MODIFICATION_MAPPING)
my_mods["Methylation"] = ('Methylation',14.01565)
my_mods
```

If you have any additional modifications in your result file(s) the parser needs to know about them, which is done via the `modifications` parameter that allows for passing a custom dictionary of modifications. 

It is usually a good idea to base this custom dictionary on `constants.SCOUT_MODIFICATION_MAPPING` and add your modifications after, as shown here for methylation.

```python
parser_result_Cas9_filtered_CSMs = parser.read_scout(
    "../../data/scout/Cas9_Filtered_CSMs.csv",
    crosslinker="DSSO",
    modifications=my_mods
)
```

You can then pass the full list of expected modifications `my_mods` via the `modifications` parameter.

```python
_ = transform.summary(parser_result_Cas9_filtered_CSMs)
```

There are several other parameters that can be set, you can read more about them here: [**docs**](https://hgb-bin-proteomics.github.io/pyXLMS/pyXLMS.parser.html#pyXLMS.parser.parser_xldbse_scout.read_scout).
